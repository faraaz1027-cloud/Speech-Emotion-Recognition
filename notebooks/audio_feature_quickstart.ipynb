This is a NEW portfolio notebook (not the original CSE424 analysis). It extracts 58 acoustic features from **one** local WAV file. It does **not** predict emotions. See the README for original team authors, historical metrics, and validation limitations.

```python
# Run from the repository root with Python 3.12:
# %pip install -r demo/requirements.txt
from pathlib import Path
from demo.features import load_audio, extract_features
import pandas as pd
```

```python
# Use any WAV recording you have permission to analyze.
WAV_PATH = Path('example.wav') # replace with a real local WAV file
assert WAV_PATH.is_file(), 'Set WAV_PATH to an existing WAV file.'
y, sr = load_audio(WAV_PATH.read_bytes())
features = extract_features(y, sr)
print('Duration:', round(len(y)/sr, 2), 'seconds')
print('Number of features:', len(features))
pd.DataFrame({'Feature': list(features), 'Value': list(features.values())})
```
